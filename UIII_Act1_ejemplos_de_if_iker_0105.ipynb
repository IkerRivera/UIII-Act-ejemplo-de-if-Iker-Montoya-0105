```python
# Ejemplo 2: Uso de if-else con operadores lógicos y relacionales
# Este ejemplo decide si una persona puede acceder a un descuento basado en su edad y si tiene una membresía.
edad = 26 # Se define la edad del usuario [12].
tiene_membresia = True # Variable booleana que indica estado [13].

if edad >= 26 and tiene_membresia: # '>=' es relacional y 'and' es un operador lógico [4, 9].
    print("Acceso al descuento de socio permitido.") # Se ejecuta si ambas condiciones son verdaderas [4].
else: # Bloque opcional para el caso contrario [1].
    print("No cumple los requisitos para el descuento.") # Se ejecuta si alguna condición falla [6].
```

```python

# Como las notas suelen requerir precisión decimal, es fundamental convertir la entrada de texto (que es lo que devuelve input) a un tipo numérico float para realizar cálculos aritméticos

# Aquí tienes el código modificado y la descripción de sus componentes:
# Ejemplo 3 modificado: Clasificación de rendimiento capturando datos
nota1 = float(input("Ingresa la primera calificación: ")) # Captura y convierte la primera nota a decimal.
nota2 = float(input("Ingresa la segunda calificación: ")) # Captura y convierte la segunda nota a decimal.

promedio = (nota1 + nota2) / 2 # Realiza la operación aritmética de suma y división.

if promedio >= 9: # Operador relacional para verificar excelencia.
    print("Sobresaliente")
elif promedio >= 7 and promedio < 9: # Uso de 'elif' y operador lógico 'and' para un rango.
    print("Notable")
elif promedio >= 5 or promedio * 10 > 50: # Uso de 'or' para validar dos condiciones alternativas.
    print("Aprobado")
else: # Bloque que se ejecuta si todas las condiciones anteriores son falsas.
    print("Reprobado")
```

```python
# Ejemplo 4: Calculadora de Categoría de Salud (IMC)
# Captura de datos con conversión a punto flotante (decimal)
peso = float(input("Ingresa tu peso en kg: ")) # [2, 4]
estatura = float(input("Ingresa tu estatura en metros (ej. 1.70): ")) # [2, 4]

# Uso de operadores lógicos y relacionales para validar la entrada
if peso <= 0 or estatura <= 0: # Operador relacional '<=' y lógico 'or' [5, 6]
    print("Error: El peso y la estatura deben ser mayores a cero.")
else:
    # Operadores aritméticos: división '/' y potencia '**' para calcular el IMC
    imc = peso / (estatura ** 2) # [7-9]
    print(f"Tu IMC es: {imc:.2f}")

    # Estructura if-elif-else con operadores relacionales y lógicos para clasificar
    if imc < 18.5: # Operador relacional [5, 6]
        print("Clasificación: Bajo peso")
    elif imc >= 18.5 and imc < 25: # Relacionales y lógico 'and' [5, 6]
        print("Clasificación: Peso normal")
    elif imc >= 25 and imc < 30: # Relacionales y lógico 'and' [5, 6]
        print("Clasificación: Sobrepeso")
    else: # Caso por defecto
        print("Clasificación: Obesidad") # [1]
```
